# Flood Frequency Analysis of NVE Discharge Data

This notebook performs hydrological extreme value analysis including:
- Time series visualization
- Seasonal pattern analysis
- Flood rose plot
- Annual maximum extraction
- GEV distribution fitting (L-moments)
- Diagnostic plots
- Return level estimation


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import genextreme
import lmoments3 as lm
import lmoments3.distr as lm_distr

plt.style.use('seaborn-v0_8')
```

```python
file_path = 'data/62.5.0-Vannføring-dogn-v1.csv'

df = pd.read_csv(file_path, sep=';', skiprows=1)
df.columns = ['time', 'discharge', 'corrected', 'quality']
df['time'] = pd.to_datetime(df['time'], utc=True)
df['discharge'] = (
    df['discharge']
    .str.replace(',', '.', regex=False)
    .astype(float)
)
df['discharge'] = pd.to_numeric(df['discharge'])
df = df.dropna(subset=['discharge'])
df = df.set_index('time')

df.head()
```

## Flood Rose

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

def flood_rose_plot(df, qth=0.95):

    df = df.copy().sort_index()
    discharge = df['discharge']

    # Threshold
    threshold = discharge.quantile(qth)

    # Peaks over threshold
    peaks = df[discharge > threshold]

    # Angle (seasonality)
    theta = 2 * np.pi * (peaks.index.dayofyear / 365.0)

    # Magnitude
    r = peaks['discharge'].values

    # ---- Plot ----
    plt.figure(figsize=(6,6))
    ax = plt.subplot(111, polar=True)

    # ONLY radial lines (no scatter)
    for t, radius in zip(theta, r):
        ax.plot([t, t], [0, radius], linewidth=1, color='blue', alpha=0.6)

    # Month labels
    month_angles = np.linspace(0, 2*np.pi, 12, endpoint=False)
    month_labels = ['Jan','Feb','Mar','Apr','May','Jun',
                    'Jul','Aug','Sep','Oct','Nov','Dec']

    ax.set_xticks(month_angles)
    ax.set_xticklabels(month_labels)

    # Orientation
    ax.set_theta_zero_location('N')
    ax.set_theta_direction(-1)

    ax.set_title(f'Flood Rose (Q > {int(qth*100)}th percentile)')

    plt.show()
```

```python
flood_rose_plot(df, qth=0.95)
```

```python
# Define threshold (global 95th percentile)
threshold = df['discharge'].quantile(0.95)

# Select flood events
floods = df[df['discharge'] > threshold]

# Convert timing to angle (day of year → radians)
theta = 2 * np.pi * floods.index.dayofyear / 365.25
r = floods['discharge'].values

plt.figure(figsize=(6,6))
ax = plt.subplot(111, polar=True)

# Plot as radial lines
for t, radius in zip(theta, r):
    ax.plot([t, t], [0, radius], color='blue', alpha=0.6)

# Optional: mark peaks
ax.scatter(theta, r, color='red', s=10)

# Formatting
ax.set_title('Flood Rose (Q > 95th percentile)')
ax.set_theta_zero_location('N')   # Jan 1 at top
ax.set_theta_direction(-1)        # clockwise

plt.show()
```

## Fit GEV Distribution (L-moments)

```python
import numpy as np
import lmoments3 as lm
from lmoments3 import distr as lm_distr

# =========================================================
# Prepare Annual Maxima (AMS)
# =========================================================
annual_max = annual_max.dropna().values

# =========================================================
# Compute L-moments
# =========================================================
lmom = lm.lmom_ratios(annual_max, nmom=4)

# =========================================================
# Fit GEV using L-moments
# =========================================================
gev_fit = lm_distr.gev.lmom_fit(lmom)

# Extract parameters
shape = gev_fit['c']     # shape (xi)
loc   = gev_fit['loc']   # location (mu)
scale = gev_fit['scale'] # scale (sigma)

print('GEV Parameters (L-moments)')
print('Shape (xi):', shape)
print('Location (mu):', loc)
print('Scale (sigma):', scale)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import lmoments3 as lm
from lmoments3 import distr as lm_distr

# =========================================================
# Clean AMS
# =========================================================
data = annual_max.dropna().values
data = data[np.isfinite(data)]
n = len(data)

# =========================================================
# L-moments fit (ascending sort for fitting)
# =========================================================
data_asc = np.sort(data)
lmom = lm.lmom_ratios(data_asc, nmom=4)
gev_fit = lm_distr.gev.lmom_fit(lmom)

c = gev_fit['c']
loc = gev_fit['loc']
scale = gev_fit['scale']

print("Shape (xi, R convention):", -c)

# =========================================================
# Observed return periods (descending data)
# =========================================================
data_desc = np.sort(data)[::-1]   # largest first
rank = np.arange(1, n + 1)       # 1 = largest
T_obs = (n + 1) / rank

# =========================================================
# Theoretical return levels
# =========================================================
T_fit = np.logspace(0.01, 3, 200)
p_fit = 1 - 1 / T_fit
rl_fit = lm_distr.gev.ppf(p_fit, c=c, loc=loc, scale=scale)

# =========================================================
# Plot
# =========================================================
plt.figure(figsize=(8,6))

plt.plot(T_fit, rl_fit, color='blue', linewidth=2, label='GEV fit')
plt.scatter(T_obs, data_desc, facecolors='none', edgecolors='black', label='Observed AMS', zorder=10)

plt.xscale('log')
plt.xlabel('Return period [years]')
plt.ylabel('Discharge [m³/s]')
plt.grid(True, which='both', linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import lmoments3 as lm
from lmoments3 import distr as lm_distr

# =========================================================
# Data
# =========================================================
data = annual_max.dropna().values
n = len(data)

# Return periods
T = np.logspace(0, 3, 100)
p = 1 - 1/T

# =========================================================
# Fit GEV (L-moments)
# =========================================================
lmom = lm.lmom_ratios(data, nmom=4)
gev_fit = lm_distr.gev.lmom_fit(lmom)

shape = gev_fit['c']
loc   = gev_fit['loc']
scale = gev_fit['scale']

# Median estimate
rl_median = lm_distr.gev.ppf(p, c=shape, loc=loc, scale=scale)

# =========================================================
# Bootstrap for 95% CI
# =========================================================
n_boot = 500
rl_boot = np.zeros((n_boot, len(T)))

for i in range(n_boot):
    sample = np.random.choice(data, size=n, replace=True)
    lmom_b = lm.lmom_ratios(sample, nmom=4)
    fit_b = lm_distr.gev.lmom_fit(lmom_b)
    
    rl_boot[i, :] = lm_distr.gev.ppf(
        p,
        c=fit_b['c'],
        loc=fit_b['loc'],
        scale=fit_b['scale']
    )

# Confidence intervals
rl_lower = np.percentile(rl_boot, 2.5, axis=0)
rl_upper = np.percentile(rl_boot, 97.5, axis=0)

# =========================================================
# Observed AMS (Weibull plotting position)
# =========================================================
sorted_data = np.sort(data)
T_obs = (n + 1) / np.arange(1, n + 1)

# =========================================================
# Plot (match your figure style)
# =========================================================
plt.figure(figsize=(8,6))

# GEV curve
plt.plot(T, rl_median, label='GEV (L-moments)', linewidth=2)

# Confidence band
plt.fill_between(
    T,
    rl_lower,
    rl_upper,
    alpha=0.25,
    label='95% CI'
)

# Observations
plt.scatter(
    T_obs,
    sorted_data,
    facecolors='none',
    edgecolors='black',
    label='Observed AMS',
    zorder=10
)

# Axis settings
plt.xscale('log')
plt.xlabel('Return period [years]')
plt.ylabel('Flood size [m³/s]')

# Grid like your figure
plt.grid(True, which='both', linestyle='--', alpha=0.5)

plt.legend()
plt.title('GEV-fitted Annual Maxima Return Period Plot')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import genextreme
from scipy.optimize import minimize
from datetime import datetime
```

```python
 
return_times = [100, 200, 1000]

# --- 1. Load discharge data (assumes first column is date, second is flow) ---
dis_d = df["discharge"]
dis_d.columns = ["date", "flow"]
```

```python
dis_d
```

```python
 
ams = dis_d.resample("YE").max().values
years = np.unique(dis_d.index.year)
alpha = 0.05
return_times = [100, 200, 1000]

# --- GEV fit (ML) ---
def gev_nll(params, data):
    c, loc, scale = params
    if scale <= 0:
        return np.inf
    return -np.sum(genextreme.logpdf(data, c, loc=loc, scale=scale))

init = [0.1, np.mean(ams), np.std(ams)]
result = minimize(gev_nll, init, args=(ams,), method="L-BFGS-B")
c_mle, loc_mle, scale_mle = result.x
shape_mle = -c_mle

# --- Bootstrap to generate uncertainty intervals ---
n_boot = 1000
rl_boot = {T: [] for T in return_times}
rng = np.random.default_rng()

for _ in range(n_boot):
    sample = rng.choice(ams, size=len(ams), replace=True)
    res_boot = minimize(gev_nll, init, args=(sample,), method="L-BFGS-B")
    if res_boot.success:
        c_b, loc_b, scale_b = res_boot.x
        shape_b = -c_b
        for T in return_times:
            rl_boot[T].append(genextreme.ppf(1-1/T, -shape_b, loc=loc_b, scale=scale_b))

print("AMS-GEV flood estimates for return periods with 95% CI:")
for T in return_times:
    rl_ml = genextreme.ppf(1-1/T, -shape_mle, loc=loc_mle, scale=scale_mle)
    ci_lo, ci_up = np.percentile(rl_boot[T], [100*alpha/2, 100*(1-alpha/2)])
    print(f"{T} years: {rl_ml:.2f} (95% CI: {ci_lo:.2f} - {ci_up:.2f}) m3/s")

# --- Plot return level curve with uncertainty bands ---
T_range = np.linspace(2, 1000, 100)
rl_vals = genextreme.ppf(1-1/T_range[:,None], -shape_mle, loc=loc_mle, scale=scale_mle)

# Compute uncertainty bands from bootstrap
rl_low = []
rl_high = []
for T in T_range:
    rl_low.append(np.percentile(rl_boot[T], 100*alpha/2))
    rl_high.append(np.percentile(rl_boot[T], 100*(1-alpha/2)))

plt.figure()
plt.plot(T_range, genextreme.ppf(1-1/T_range, -shape_mle, loc=loc_mle, scale=scale_mle), label="Return level (ML)")
plt.fill_between(T_range, rl_low, rl_high, color="lightblue", alpha=0.4, label="95% CI")
plt.xscale("log")
plt.xlabel("Return period (years)")
plt.ylabel("Flood size (m3/s)")
plt.title("AMS-GEV Return Level with 95% CI")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import genextreme
from scipy.interpolate import interp1d

# T_range for plotting (continuous)
T_range = np.linspace(2, 1000, 100)

# ML return level curve
rl_ml_curve = genextreme.ppf(1 - 1/T_range, -shape_mle, loc=loc_mle, scale=scale_mle)

# For uncertainty bands, interpolate from the bootstrap results at the discrete return periods
T_boot = np.array(return_times)
rl_lo_boot = np.array([np.percentile(rl_boot[T], 100*alpha/2) for T in return_times])
rl_hi_boot = np.array([np.percentile(rl_boot[T], 100*(1-alpha/2)) for T in return_times])

# Interpolate to match T_range
interp_lo = interp1d(T_boot, rl_lo_boot, kind='linear', fill_value="extrapolate")
interp_hi = interp1d(T_boot, rl_hi_boot, kind='linear', fill_value="extrapolate")

rl_low_curve = interp_lo(T_range)
rl_high_curve = interp_hi(T_range)

# Plot
plt.figure()
plt.plot(T_range, rl_ml_curve, label="Return level (ML)")
plt.fill_between(T_range, rl_low_curve, rl_high_curve, color="lightblue", alpha=0.4, label="95% CI")
plt.xscale("log")
plt.xlabel("Return period (years)")
plt.ylabel("Flood size (m3/s)")
plt.title("AMS-GEV Return Level with 95% CI")
plt.legend()
plt.grid(True)
plt.show()
```
